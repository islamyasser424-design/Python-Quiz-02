```python
orders_cairo = [1200.0, "INVALID_DATA", 850.0, 1500.0]
orders_alex = [950.0, 450.0, 1100.0, 300.0]
delivery_fees = [50.0, 40.0, 60.0, 50.0]

valid_totals = []

print("=== BRANCH AUDIT REPORT ===")

for i in range(4):

    cairo = orders_cairo[i]
    alex = orders_alex[i]
    fee = delivery_fees[i]

    if isinstance(cairo, (int, float)):
        cairo_total = cairo + fee
        valid_totals.append(cairo_total)
        cairo_result = f"EGP {cairo_total:,.2f}"
    else:
        print(f"Warning: Invalid Cairo order at Index {i}. Skipped.")
        cairo_result = "Skipped"

    alex_total = alex + fee
    valid_totals.append(alex_total)
    alex_result = f"EGP {alex_total:,.2f}"

    print(f"Order #{i + 1}: Cairo = {cairo_result} | Alex = {alex_result}")

print("===========================")
print("Total Valid Orders :", len(valid_totals))
print(f"Grand Total Revenue: EGP {sum(valid_totals):,.2f}")
print(f"Highest Single Order: EGP {max(valid_totals):,.2f}")
print(f"Lowest Single Order : EGP {min(valid_totals):,.2f}")
```
